```python
Active_Transport = ['Walking','Bike']
condition=[df["transport"].isin (Active_Transport)]
result=["Active_Transport"]
df["dicotomical_transport"]=np.select(condition, result, default="NotActive_Transport")
```

```python
Obesity = ['Obesity_Type_I', 'Obesity_Type_II', 'Obesity_Type_III']
condition=[df["obesity_level"].isin (Obesity)]
result=["Obesity"]
df["dicotomical_Obesity"]=np.select(condition, result, default="obesity_level")
```

```python
high_frequency = ['Frequently', 'Always']
condition = [df["snack_freq"].isin(high_frequency)]
result = ["High"]
df["dicotomical_snack_freq"] = np.select(condition, result, default="Low")
```

```python
condition = [df["alcohol_freq"] != "no"]
result = ["Yes"]
df["dicotomical_alcohol"] = np.select(condition, result, default="No")
```

```python
mappings = {
    'snack_freq': {'no': 0, 'Sometimes': 1, 'Frequently': 2, 'Always': 3},
    'alcohol_freq': {'no': 0, 'Sometimes': 1, 'Frequently': 2, 'Always': 3},
    'obesity_level': {
        'Insufficient_Weight': 0, 'Normal_Weight': 1, 
        'Overweight_Level_I': 2, 'Overweight_Level_II': 3, 
        'Obesity_Type_I': 4, 'Obesity_Type_II': 5, 'Obesity_Type_III': 6
    },
    'dicotomical_snack_freq': {'Low': 0, 'High': 1, 'Baja o Nula': 0, 'Alta': 1},
    'dicotomical_alcohol': {'No': 0, 'Yes': 1},
    'dicotomical_Obesity': {'obesity_level': 0, 'Obesity': 1}
}

for col, mapping in mappings.items():
    df[f"{col}_encoded"] = df[col].map(mapping)
```

```python
Numerical_Features_NOTTarget = df.select_dtypes(include=['float64', 'int64']).drop(columns=[ "Number_Target"], errors='ignore').columns.tolist()
```

```python
X =df[Numerical_Features_NOTTarget]
y = df["Number_Target"]
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.20, random_state = 1)
```

```python
from sklearn.neighbors import KNeighborsClassifier
knn = KNeighborsClassifier(n_neighbors = 5, metric = 'euclidean')
knn.fit(X_train, y_train)
```

```python
from sklearn.metrics import confusion_matrix
import seaborn as sns

# Call a method predict by using an object classifier 'cls_svm'
y_predict = knn.predict(X_test)

# Calculate cm by calling a method named as 'confusion_matrix'
cm = confusion_matrix(y_test, y_predict)

# Call a method heatmap() to plot confusion matrix
sns.heatmap(cm, annot = True)
```

```python
from sklearn.metrics import classification_report

# Display the classification report
print(classification_report(y_test, y_predict))
```

```python
import numpy as np
neighbors = np.arange(1, 11)                  # number of neighbors
train_accuracy = np.zeros(len(neighbors))     # Declare and initialise the matrix
test_accuracy = np.zeros(len(neighbors))      # Declare and initialise the matrix
```

```python
for i, k in enumerate(neighbors):                          # for loop that checks the model for neighbor values 1, 2, 3, ..., 10
    kNN = KNeighborsClassifier(n_neighbors = k)            # Initialise an object knn using KNeighborsClassifier method

    # Fit the model
    kNN.fit(X_train, y_train)                              # Call fit method to implement the ML KNeighborsClassifier model

    # Compute accuracy on the training set
    train_accuracy[i] = kNN.score(X_train, y_train)        # Save the score value in the train_accuracy array

    # Compute accuracy on the test set
    test_accuracy[i] = kNN.score(X_test, y_test)           # Save the score value in the train_accuracy array
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize = (10, 6))
plt.title('KNN accuracy with varying number of neighbors', fontsize = 20)
plt.plot(neighbors, test_accuracy, label = 'Testing Accuracy')
plt.plot(neighbors, train_accuracy, label = 'Training accuracy')
plt.legend(prop={'size': 20})
plt.xlabel('Number of neighbors', fontsize = 20)
plt.ylabel('Accuracy', fontsize = 20)
plt.xticks(fontsize = 20)
plt.yticks(fontsize = 20)
plt.show()
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
scaler.fit(X_train)
X_train_scaled = scaler.transform(X_train)
X_test_scaled  = scaler.transform(X_test)
X_scaled = scaler.transform(X)
print(X_train_scaled, '\t', X_test_scaled, '\t', X_scaled)
```

```python
from sklearn.decomposition import PCA
pca = PCA(n_components = 5)
X_pca = pca.fit_transform(X_scaled)
print("Original shape: {}".format(str(X_scaled.shape)))
print("Reduced shape: {}".format(str(X_pca.shape)))
```
